1.Multiplication Table

```python
class MultiplicationTable:

    """
    MultiplicationTable class handles printing the multiplication
    table of a given number.

    Attributes:
        number (int): The number whose multiplication table
                      will be printed.
    """

    def __init__(self, number: int) :
        """
        Initialize the MultiplicationTable object.

        Parameters:
            number (int): The number for the multiplication table.
        """

        self.number = number

    def generate_table(self) :
        """
        Generate and print the multiplication table.

        Returns:
            None: This method only prints the multiplication table.
        """

        for i in range(1, 11):
            print(f"{self.number} x {i} = {self.number * i}")



table = MultiplicationTable(5)

table.generate_table()
```

Task 2 — Twin Primes Less Than 1000

```python
class TwinPrimes:
    """
    TwinPrimes class handles finding twin prime numbers
    within a given limit.

    Attributes:
        limit (int): The upper limit for finding twin primes.
    """

    def __init__(self, limit: int) :
        """
        Initialize the TwinPrimes object.

        Parameters:
            limit (int): The upper limit for the search.
        """

        self.limit = limit

    def is_prime(self, number: int) :
        """
        Check whether a number is prime.

        Parameters:
            number (int): The number to check.

        Returns:
            bool: True if the number is prime, otherwise False.
        """

        if number < 2:
            return False

        for i in range(2, number):
            if number % i == 0:
                return False

        return True

    def find_twin_primes(self) :
        """
        Find all twin prime pairs below the given limit.

        Returns:
            list[tuple[int, int]]: A list containing twin prime pairs.
        """

        twin_primes = []

        for number in range(3, self.limit, 2):

            if self.is_prime(number) and self.is_prime(number + 2):
                twin_primes.append((number, number + 2))

        return twin_primes


twins = TwinPrimes(1000)


result = twins.find_twin_primes()


for pair in result:
    print(pair)
```

Task 3 — Prime Factors

```python
class PrimeFactors:
    """
    PrimeFactors class handles finding the prime factors
    of a given number.

    Attributes:
        number (int): The number whose prime factors
                      will be found.
    """

    def __init__(self, number: int) :
        """
        Initialize the PrimeFactors object.

        Parameters:
            number (int): The number to find its prime factors.
        """

        self.number = number

    def find_factors(self) :
        """
        Find the prime factors of the number.

        Returns:
            list[int]: A list containing the prime factors.
        """

        number = self.number
        factors = []
        divisor = 2

        while number > 1:

            if number % divisor == 0:
                factors.append(divisor)
                number = number // divisor

            else:
                divisor += 1

        return factors



factors = PrimeFactors(56)


result = factors.find_factors()


print(result)
```

4 — Decimal to Binary

```python
class DecimalToBinary:

    """
    DecimalToBinary class handles converting a decimal number
    into its binary representation.

    Attributes:
        number (int): The decimal number to be converted.
    """

    def __init__(self, number: int) :
        """
        Initialize the DecimalToBinary object.

        Parameters:
            number (int): The decimal number to convert.
        """

        self.number = number

    def convert(self) :
        """
        Convert the decimal number to binary.

        Returns:
            str: The binary representation of the number.
        """

        number = self.number
        binary = ""

        if number == 0:
            return "0"

        while number > 0:
            remainder = number % 2
            binary = str(remainder) + binary
            number = number // 2

        return binary



binary_number = DecimalToBinary(10)


result = binary_number.convert()


print(result)
```

Task 5 — Perfect Numbers in a Range

```python
class PerfectNumbers:
    """
    PerfectNumbers class handles finding perfect numbers
    within a given range.

    Attributes:
        start (int): The starting value of the range.
        end (int): The ending value of the range.
    """

    def __init__(self, start: int, end: int) :
        """
        Initialize the PerfectNumbers object.

        Parameters:
            start (int): The starting value of the range.
            end (int): The ending value of the range.
        """

        self.start = start
        self.end = end

    def find_perfect_numbers(self) :
        """
        Find all perfect numbers within the given range.

        Returns:
            list[int]: A list containing all perfect numbers.
        """

        perfect_numbers = []

        for number in range(self.start, self.end + 1):

            total = 0

            for divisor in range(1, number):

                if number % divisor == 0:
                    total += divisor

            if total == number:
                perfect_numbers.append(number)

        return perfect_numbers



perfect = PerfectNumbers(1, 1000)

result = perfect.find_perfect_numbers()


print(result)
```
