2. Write a program to print twin primes less than 1000. If two
consecutive odd numbers are both prime then they are known as
twin primes

```python
def twin_primes(limit: int) :
    """
    Print all twin prime pairs less than a given limit.

    Parameters:
        limit (int): The upper limit for finding twin primes.

    Returns:
        None

    Example:
        >>> twin_primes(10)
        (3, 5)
        (5, 7)
    """

    for number in range(3, limit, 2):

        # Check if number is prime
        is_prime1 = True

        for i in range(2, number):
            if number % i == 0:
                is_prime1 = False
                break

        # Check if number + 2 is prime
        is_prime2 = True

        for i in range(2, number + 2):
            if (number + 2) % i == 0:
                is_prime2 = False
                break

        if is_prime1 and is_prime2:
            print(number, number + 2)


twin_primes(1000)
```

3. Write a program to find out the prime factors of a number.
Example: prime factors of 56 - 2, 2, 2, 7

```python
def prime_factors(number: int) :
    """
    Print the prime factors of a given number.

    Parameters:
        number (int): The number to find its prime factors.

    Returns:
        None

    Example:
        >>> prime_factors(56)
        2
        2
        2
        7
    """

    divisor = 2

    while number > 1:

        if number % divisor == 0:
            print(divisor)
            number = number // divisor

        else:
            divisor += 1


prime_factors(56)
```

4.Write a function that converts a decimal number to binary
number

```python
def decimal_to_binary(number: int) :
    """
    Convert a decimal number to its binary representation.

    Parameters:
        number (int): The decimal number to convert.

    Returns:
        str: The binary representation of the number.

    Example:
        >>> decimal_to_binary(10)
        '1010'
    """

    binary = ""

    if number == 0:
        return "0"

    while number > 0:
        remainder = number % 2
        binary = str(remainder) + binary
        number = number // 2

    return binary


print(decimal_to_binary(10))
```

5.A number is called perfect if the sum of proper divisors of that
number is equal to the number. For example 28 is perfect
number, since 1+2+4+7+14=28. Write a program to print all the
perfect numbers in a given range

```python
def perfect_numbers(start: int, end: int) :
    """
    Print all perfect numbers within a given range.

    Parameters:
        start (int): The starting value of the range.
        end (int): The ending value of the range.

    Returns:
        None

    Example:
        >>> perfect_numbers(1, 1000)
        6
        28
        496
    """

    for number in range(start, end + 1):

        total = 0

        for divisor in range(1, number):
            if number % divisor == 0:
                total += divisor

        if total == number:
            print(number)


perfect_numbers(1, 1000)
```
